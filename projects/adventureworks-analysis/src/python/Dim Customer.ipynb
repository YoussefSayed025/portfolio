```python
df['Store Name'] = df['Store Name'].fillna('Unknown')
```

```python
df['customer type'] = df['Store Name'].apply(
    lambda x: 'Store' if x != 'Unknown' else 'Individual'  )
```

```python
df['Person Type'] = df['Person Type'].fillna('Store Without contact')
```

```python
mapping = {
    'IN': 'Individual',
    'SC': 'Store Contact',
}

df['Person Type'] = df['Person Type'].map(mapping)
```

```python
df['First Name'] = df['First Name'].fillna('Unknown')
df['Last Name'] = df['Last Name'].fillna('Customer')
```
