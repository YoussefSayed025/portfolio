```python
import pandas as pd
import numpy as np


df = pd.read_csv("CSV Files\Dim_Product.csv")


df.dropna(subset='Subcategory ID',inplace=True)
```

```python
df.info()
```

```python
df.columns
```

```python
df['Subcategory ID']=df['Subcategory ID'].astype('int64')
```

```python
df['Is Manufactured']=df['Is Manufactured'].replace({0:False,1:True})
```

```python
# Convert numeric sizes to labels
size = pd.to_numeric(df['Product Size'], errors='coerce')
conditions = [
    size <= 20,
    size <= 40,
    size <= 50,
    size <= 60,
    size > 60
]
choices = ['S', 'M', 'L', 'X', 'XL']
df['Product Size'] = np.select(conditions, choices, default='XL')
```

```python

df.loc[ (
        df["Product Name"]
        .str.lower()
        .str.endswith(
            " s",
        )
    ),
    "Product Size",
] = "S"
df.loc[ (
        df["Product Name"]
        .str.lower()
        .str.endswith(
            " m",
        )
    ),
    "Product Size",
] = "M"
df.loc[ (
        df["Product Name"]
        .str.lower()
        .str.endswith(
            " l",
        )
    ),
    "Product Size",
] = "L"
df.loc[ (
        df["Product Name"]
        .str.lower()
        .str.endswith(
            " xl",
        )
    ),
    "Product Size",
] = "XL"
```

```python
df["Size Unit Measure"] = df["Size Unit Measure"].fillna(df["Size Unit Measure"].mode()[0])
df["Weight Unit Measure"] = df["Weight Unit Measure"].fillna( df["Weight Unit Measure"].mode()[0])
df["Product Class"] = df["Product Class"].fillna( df["Product Class"].mode()[0])
df["Product Style"] = df["Product Style"].fillna( df["Product Style"].mode()[0])
df["Product Color"] = df["Product Color"].fillna( df["Product Color"].mode()[0])
```

```python
df.drop(columns=['Sell End Date','Sell Start Date'],inplace=True)
```

```python
def categorize_days(days):
    if pd.isna(days):
        return None
    elif days == 0:
        return "Same Day"
    elif days <= 7:
        return "Within a Week"
    elif days <= 30:
        return "Within a Month"
    elif days <= 90:
        return "Within 3 Months"
    elif days <= 365:
        return "Within a Year"
    else:
        return "Over a Year"


df['Days To Manufacture']=df['Days To Manufacture'].apply(categorize_days)
```
